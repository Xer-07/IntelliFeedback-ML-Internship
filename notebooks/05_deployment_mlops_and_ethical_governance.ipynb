## 1. PII Redaction Demonstration (GDPR / CCPA Compliance)
Stateless regular expressions mask customer identifiers before text vectorization.

````python
def scrub_pii(text: str) -> str:
    text = re.sub(r'[\w\.-]+@[\w\.-]+', '[REDACTED_EMAIL]', text)
    text = re.sub(r'\b\d{3}[-.]?\d{3,4}\b', '[REDACTED_PHONE]', text)
    return text

raw_sample = "Please contact john.doe@example.com or 555-0199 regarding broken order #9872."
scrubbed = scrub_pii(raw_sample)
print(f"Raw     : {raw_sample}")
print(f"Scrubbed: {scrubbed}")
````
